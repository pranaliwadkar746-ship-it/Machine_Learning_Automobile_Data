```python
#Task 1: Prepare Data for Model Building


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

Automobile = pd.read_csv('Automobile_data.csv')
```

```python
## Standardize the data in the columns
Automobile["normalized-losses"].replace
```

```python
# check for null values

Automobile.isnull().sum()
```

```python
# Fill the missing values
Automobile["normalized-losses"] = pd.to_numeric(
    Automobile["normalized-losses"],
    errors="coerce"
)

m=Automobile["normalized-losses"].mean()
print("Mean of normalized-losses : ",m)
Automobile["normalized-losses"]=Automobile["normalized-losses"].fillna(m,inplace=True)
```

```python
Automobile.isnull().sum()
```

```python
Automobile["horsepower"] = Automobile["horsepower"].replace("?", np.nan)
```

```python
#Do the same for Horse-power Column

#replace ? with null values in "horsepower" column permanently
#change the datatype of Automobile["horsepower"] column from object to float
#permanently
Automobile["horsepower"]=Automobile["horsepower"].astype("float")
```

```python
"?" in Automobile["horsepower"].values
```

```python
Automobile.isnull().sum()
```

```python
M=Automobile["horsepower"].mean()
print("Mean of horsepower : ",m)
Automobile["horsepower"]=Automobile["horsepower"].fillna(m,inplace=True)
```

```python
#separate numerical type column and object type column and store in new dataset

Automobile_num=Automobile.select_dtypes(["int64","float64"])#hold to int and float type data 
Automobile_cat=Automobile.select_dtypes(object)#hold object type data
```

```python
#Encode the Categorical Columns

from sklearn.preprocessing import LabelEncoder
for col in Automobile_cat :
    le=LabelEncoder() #create the object of LabelEncoder class
    Automobile_cat[col]=le.fit_transform(Automobile_cat[col])
```

```python
Automobile_cat.head()
```

```python
#Concatenation of both dataset df_num and df_cat and hold new dataset df_new

Automobile_new=pd.concat([Automobile_num,Automobile_cat],axis=1) #axis=1 column wise
Automobile_new.head()
```

```python
#Select independent means input variables(X) and target means dependent variable(Y)

X=Automobile_new.drop("price",axis=1) #apart from price, all columns hold in X variable
Y=Automobile_new["price"] #output variable
```

```python
#Train test split 70%-30%

from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,
                                              random_state=1)
```

```python
#Check the shape of test and train data

X_train.shape,X_test.shape
```

```python
#Apply StandardScaler on X_train and X_test

from sklearn.preprocessing import StandardScaler
ss=StandardScaler()
X_train=ss.fit_transform(X_train)
X_test=ss.transform(X_test)
```

```python
X_train
```

```python
#Task 2: Build and Train Linear Regression Model

#Import Linear Regression

from sklearn.linear_model import LinearRegression
```

```python
#Create Model Object

lr=LinearRegression()
```

```python
#Train the Model with 70% data

lr.fit(X_train,Y_train)
```

```python
#Task 3: Predict Vehicle Prices

#Generate Predictions

y_pred = lr.predict(X_test)
```

```python
#View Predicted Prices

y_pred[:10]
```

```python
print('b0 (intercept) : ',lr.intercept_)
print('b1 (slope) : ',lr.coef_[0])
print('b2 (bedroom) : ',lr.coef_[1])
```

```python
#Calculate the R² Score

r2 = r2_score(Y_test, y_pred)
print("R² Score:", r2)
```
